```python
import math

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
math.pi
```

```python
theta = np.linspace(0, math.pi, num=16)
```

```python
l_top = 2 * np.sin(theta / 2)
```

```python
l_side = np.sqrt((math.pi - theta) ** 2 + 4)
```

```python
#plt.plot(theta, l_top)
#plt.plot(theta, l_side)
plt.plot(theta, l_top + l_side)
```

```python
df = pd.DataFrame({"theta": theta, "top": l_top, "side": l_side, "total": l_top + l_side})
df
```

# 2026-08-14: Camp Algebra

Camp Algebra held a competition in which campers were split into two teams: the Aces and the Ciphers. Various members of these teams competed in a triathlon of chess, basketball, and baseball. The results are shown below. To commemorate the event, the camp counselors designed a cuboidal trophy with dimensions that cleverly related to the results of the three events. Then, being Camp Algebra after all, they computed the square of the shortest path along the trophy’s surface from one corner to the opposite corner. After decoding that number, namely, what camp mascot did they carve into the trophy?

## Part 1 (Chess)

Occupatus, playing as white for the Aces, had Castor, playing as black for the Ciphers, on the ropes. But the industrious Castor clearly gave a dam. With what next move did Castor turn the tables?

```python
df = pd.read_csv("data-chess.csv")
df
```

```python
game = pd.concat([df["X.1"], df["X.2"]], ignore_index=True).to_frame(name="move")
game["occupatus"] = pd.concat([df["Occupatus"], df["Occupatus.1"]], ignore_index=True)
game["castor"] = pd.concat([df["Castor"], df["Castor.1"]], ignore_index=True)
game.loc[23, "castor"] = ''
game
```

```python
def write_pgn(df, filename):
    with open(filename, "w", encoding="utf-8") as fh:
        for row in df.itertuples():
            fh.write(f"{row.move}.{row.occupatus} {row.castor}\n")

write_pgn(game, "game.pgn")
```

Loading the PGN file into `xboard`, we get this:

![](game-position.png)

From here, the best move in [algebraic chess notation](https://en.wikipedia.org/wiki/Algebraic_notation_(chess)) is

    Bb4

which puts the white king in check and I believe forces a mate in 1 for black.
